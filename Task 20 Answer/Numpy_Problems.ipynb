### 1- Write a NumPy program to create a 5x5 array with random values and find the index of the maximum value in each row ?

```python
# axis=1 searches across each row. The result is a column index.
# argmax returns the position of the largest value.
arr = np.random.rand(5, 5)
indices = np.argmax(arr, axis=1)
print('Array:\n', arr)
print('Maximum indices in each row:', indices)
```

### 2- Write a NumPy program to create a 5x5 array with random values and find the index of the minimum value in each row

```python
# argmin returns the position of the smallest value.
arr = np.random.rand(5, 5)
indices = np.argmin(arr, axis=1)
print('Array:\n', arr)
print('Minimum indices in each row:', indices)
```

### 4-Replace all values in a 1D array with the mean of the array.

### 6-Write a NumPy program to create a 4x4 array with random values and compute the eigenvalues and eigenvectors

```python
# Each column of eigenvectors belongs to the matching eigenvalue.
arr = np.random.rand(4, 4)
eigenvalues, eigenvectors = np.linalg.eig(arr)
print('Array:\n', arr)
print('Eigenvalues:', eigenvalues)
print('Eigenvectors:\n', eigenvectors)
print('A times V equals V times eigenvalues:',
      np.allclose(np.dot(arr, eigenvectors), eigenvectors * eigenvalues))
```

### 7-Create a random 3x3 matrix and replace all values greater than 0.5 with 1 and all others with 0.



```python
arr = np.random.rand(3, 3)
print('Original:\n', arr)
# where chooses 1 when the condition is true and 0 otherwise.
result = np.where(arr > 0.5, 1, 0)
print('Result:\n', result)
```

### 14-Calculate the mean absolute deviation of a 1D array.



```python
arr = np.array([2, 4, 6, 8])
mean_value = np.mean(arr)
deviations = np.abs(arr - mean_value)
print('Mean:', mean_value)
print('Absolute deviations:', deviations)
print('Mean absolute deviation:', np.mean(deviations))
```

### 17-Create a 1D array and set the values between the 25th and 75th percentile to 0.



```python
# Here "between" includes the two percentile boundaries.
arr = np.arange(1, 10, dtype=float)
q25, q75 = np.percentile(arr, [25, 75])
arr[(arr >= q25) & (arr <= q75)] = 0
print('25th percentile:', q25)
print('75th percentile:', q75)
print('Result:', arr)
```

### 25- How to convert a float (32 bits) array into an integer (32 bits) ?

```python
arr = np.array([1.2, 2.8, -3.7], dtype=np.float32)
# astype creates a converted array and truncates towards zero.
result = arr.astype(np.int32)
print(result)
print('Data type:', result.dtype)
```
